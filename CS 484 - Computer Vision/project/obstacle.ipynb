```python
#OPTİMAL VERSİYON DOKUNMA BURAYA


import cv2
import numpy as np
from matplotlib import pyplot as plt

def evaluate_detector(detector, image):
    keypoints = detector.detect(image)
    return len(keypoints)  # You can customize this evaluation function as needed

def find_optimal_params(image):
    # Define the ranges for parameter tuning
    min_thresholds = [50, 100, 150]
    max_thresholds = [200, 255]
    min_areas = [20, 50, 100, 150]
    min_circularities = [0.1, 0.3, 0.5]
    min_convexities = [0.3, 0.5, 0.7]
    min_inertia_ratios = [0.01, 0.1, 0.3]

    best_score = -1
    best_params = None
    best_keypoints = None

    for min_threshold in min_thresholds:
        for max_threshold in max_thresholds:
            for min_area in min_areas:
                for min_circularity in min_circularities:
                    for min_convexity in min_convexities:
                        for min_inertia_ratio in min_inertia_ratios:
                            # Set up SimpleBlobDetector parameters
                            params = cv2.SimpleBlobDetector_Params()
                            params.minThreshold = min_threshold
                            params.maxThreshold = max_threshold
                            params.filterByArea = True
                            params.minArea = min_area
                            params.filterByCircularity = True
                            params.minCircularity = min_circularity
                            params.filterByConvexity = True
                            params.minConvexity = min_convexity
                            params.filterByInertia = True
                            params.minInertiaRatio = min_inertia_ratio

                            # Create a detector with the parameters
                            ver = (cv2.__version__).split('.')
                            if int(ver[0]) < 3:
                                detector = cv2.SimpleBlobDetector(params)
                            else:
                                detector = cv2.SimpleBlobDetector_create(params)

                            # Evaluate the detector
                            score = evaluate_detector(detector, image)
                            if score > best_score:
                                best_score = score
                                best_params = {
                                    "minThreshold": min_threshold,
                                    "maxThreshold": max_threshold,
                                    "minArea": min_area,
                                    "minCircularity": min_circularity,
                                    "minConvexity": min_convexity,
                                    "minInertiaRatio": min_inertia_ratio
                                }
                                best_keypoints = detector.detect(image)

    return best_params, best_keypoints

# Read the image (assuming you've uploaded 'blob.jpg' to Colab's file system)
im = cv2.imread("blob3.jpeg", cv2.IMREAD_GRAYSCALE)

if im is None:
    print("Error: Image not loaded. Please check the file path.")
else:
    # Find optimal parameters for the image
    optimal_params, keypoints = find_optimal_params(im)

    # Print the optimal parameters found
    print(f"Optimal parameters for the image: {optimal_params}")

    # Draw detected blobs with optimal parameters
    im_with_keypoints = cv2.drawKeypoints(
        im, keypoints, np.array([]), (0, 0, 255), cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
    )

    # Display the image with keypoints using matplotlib
    plt.imshow(im_with_keypoints, cmap="gray")
    plt.title("Keypoints with Optimal Parameters")
    plt.axis("off")
    plt.show()
```

```python
# 1. VERSİYON DOKUNMA

import cv2
import numpy as np
from matplotlib import pyplot as plt

# Read the image (assuming you've uploaded 'blob.jpg' to Colab's file system)
im = cv2.imread("blob.jpg", cv2.IMREAD_GRAYSCALE)

# Setup SimpleBlobDetector parameters
params = cv2.SimpleBlobDetector_Params()

# Change thresholds
params.minThreshold = 100
params.maxThreshold = 255

# Filter by Area
params.filterByArea = True
params.minArea = 60

# Filter by Circularity
params.filterByCircularity = True
params.minCircularity = 0.4

# Filter by Convexity
params.filterByConvexity = True
params.minConvexity = 0.8

# Filter by Inertia
params.filterByInertia = True
params.minInertiaRatio = 0.01

# Create a detector with the parameters
ver = (cv2.__version__).split('.')
if int(ver[0]) < 3:
    detector = cv2.SimpleBlobDetector(params)
else:
    detector = cv2.SimpleBlobDetector_create(params)

# Detect blobs
keypoints = detector.detect(im)

# Draw detected blobs as rectangles
im_with_keypoints = im.copy()
for kp in keypoints:
    x, y = int(kp.pt[0]), int(kp.pt[1])
    size = int(kp.size / 2)
    top_left = (x - size, y - size)
    bottom_right = (x + size, y + size)
    cv2.rectangle(im_with_keypoints, top_left, bottom_right, (0, 0, 255), thickness=2)

# Display the image with keypoints using matplotlib
plt.imshow(im_with_keypoints, cmap="gray")
plt.title("Keypoints (Rectangles)")
plt.axis("off")
plt.show()
```
